Python Operator:
* Aritmetic   Operator
* Assignment Operator
* Comparison Operator
* Logical Operator
*  Membership Operator

```python
# Aritmetic Operator                          --> used with numeric values to perform common mathematics
x = 3
y = 5
print(x+y, x-y, x*y, x/y, x%y, x**y)
```

```python
# Assignment Operator (=, +=, -=, *=, /=, %=, **=)                   --> used to assign values to variable
x ,y = 3 ,8
# print(x=y, x += y ,x -= y, x *= y, x /= y, x %= y, x **= y)
x += y            # x = x + y
print(x)
```

```python
# Comparison Operator  ( ==, != , <, >, <=, >=)                         --> used to compare two values
x ,y = 3 ,8
print(x==y, x!=y, x<y, x>y, x<=y, x>=y)
```

```python
# Logical Operator   ( ==, != , <, >, <=, >=)                         --> used to combined conditional statements
print(x and y, x or y, not y)
# (x <y) and (x != y)
not((x <y) or (x == y))
```

```python
# Membership Operator ( ==, != , <, >, <=, >=)                         --> used to test if sequence is presented in an object
#print(x in y, x not in y)
x = [ 1,2,3,4,5]
y = 2
y in x
```
